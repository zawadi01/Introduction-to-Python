````python
import requests
````

````python
#import the raw data with the link of the github repository and store it with the variable name
school_url = ("https://raw.githubusercontent.com/zawadi01/supermarketdata/refs/heads/main/MOCK_DATA%20(1).json")
````

````python
#save the json file in a container
assignment = requests.get(school_url)
````

````python
#check the status code of the imported data
assignment.status_code
````

````python
#view the json file
assignment.json()
````

````python
#convert the json into a data type recognized by python
assignment_data = assignment.json()
````

````python
#check the data type
print(type(assignment_data))
````

````python
#import pandas library to deal with the dataset
import pandas as pd
````

````python
#store the list in a dataframe and convert it into a csv file
df = pd.DataFrame(assignment_data)

# Save the DataFrame to a CSV file
df.to_csv('students_data.csv', index=False)
````
